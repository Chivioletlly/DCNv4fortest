```python
# 使用 subprocess 运行（推荐，可以控制输出）
import subprocess

input_dir = "images5000/degradation"
bg_dir = "images5000/resized5000_512"
batch_size = "1"
epochs = "1"


# 运行训练并捕获输出
process = subprocess.Popen([
    "python", "train_general_decomp.py",
    "--input_dir", input_dir,
    "--batch_size", batch_size,
    "--epochs", epochs
], stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)

# 实时输出训练日志
for line in process.stdout:
    print(line, end='')

process.wait()
```
